```python
# =========================
# CELL 1 — Import Library & Load Dataset
# =========================

import pandas as pd
import numpy as np
import os

# Optional display setting
pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)

# =========================
# PATH DATASET
# =========================
# Sesuaikan kalau folder kamu beda
BASE_PATH = r"C:\Users\Gusti Jogish\Downloads\Gammafest\dataset"

TRAIN_PATH = os.path.join(BASE_PATH, "train.csv")
TEST_PATH = os.path.join(BASE_PATH, "test.csv")
SAMPLE_PATH = os.path.join(BASE_PATH, "sample submission.csv")

# =========================
# LOAD DATA
# =========================
train = pd.read_csv(TRAIN_PATH)
test = pd.read_csv(TEST_PATH)
sample_submission = pd.read_csv(SAMPLE_PATH)

# =========================
# BASIC CHECK
# =========================
print("Train shape:", train.shape)
print("Test shape:", test.shape)
print("Sample submission shape:", sample_submission.shape)

print("\nTrain columns:")
print(train.columns.tolist())

print("\nTest columns:")
print(test.columns.tolist())

print("\nSample submission columns:")
print(sample_submission.columns.tolist())

display(train.head())
display(test.head())
display(sample_submission.head())

# =========================
# MISSING VALUE CHECK
# =========================
print("\n=== Missing Values in Train ===")
display(
    train.isna().sum()
    .sort_values(ascending=False)
    .to_frame("missing_count")
    .assign(missing_pct=lambda x: x["missing_count"] / len(train) * 100)
    .query("missing_count > 0")
)

print("\n=== Missing Values in Test ===")
display(
    test.isna().sum()
    .sort_values(ascending=False)
    .to_frame("missing_count")
    .assign(missing_pct=lambda x: x["missing_count"] / len(test) * 100)
    .query("missing_count > 0")
)
```

```python
# =========================
# CELL 2 — Date Parsing, Modern Era Filter, and Column Groups
# =========================

# Copy original data for safe rollback
train_raw = train.copy()
test_raw = test.copy()

# =========================
# DATE PARSING
# =========================
train["date"] = pd.to_datetime(train["date"])
test["date"] = pd.to_datetime(test["date"])

train["year"] = train["date"].dt.year
test["year"] = test["date"].dt.year

# =========================
# MODERN ERA FILTER
# =========================
MODERN_YEAR = 1990

train = train[train["year"] >= MODERN_YEAR].copy()

# Jangan filter test berdasarkan year, karena semua test tetap harus diprediksi
# Tapi kita cek range-nya saja

# =========================
# SORT BY TIMELINE
# =========================
train = train.sort_values(["date", "match_id", "team"]).reset_index(drop=True)
test = test.sort_values(["date", "match_id", "team"]).reset_index(drop=True)

# =========================
# BASIC INFO AFTER FILTER
# =========================
print("After modern era filter")
print("Train shape:", train.shape)
print("Test shape:", test.shape)

print("\nTrain date range:", train["date"].min(), "to", train["date"].max())
print("Test date range :", test["date"].min(), "to", test["date"].max())

# =========================
# TARGET / ID COLUMNS
# =========================
TARGET_COLS = ["team_goals", "opp_goals"]

ID_COLS = [
    "id",
    "match_id",
    "date"
]

# =========================
# CATEGORICAL COLUMNS
# =========================
CAT_COLS = [
    "gender",
    "team",
    "opponent",
    "tournament",
    "venue_country",
    "confederation_team",
    "confederation_opp"
]

# Only keep categorical columns that actually exist
CAT_COLS = [col for col in CAT_COLS if col in train.columns and col in test.columns]

# =========================
# STATIC NUMERIC COLUMNS
# These are available in both train and test
# =========================
STATIC_NUM_COLS = [
    "is_home",
    "neutral",
    "population_team",
    "population_opp",
    "gdp_per_capita_team",
    "gdp_per_capita_opp",
    "altitude_venue",
    "distance_travel_team",
    "distance_travel_opp",
    "temperature_venue"
]

STATIC_NUM_COLS = [col for col in STATIC_NUM_COLS if col in train.columns and col in test.columns]

print("\nCategorical columns:")
print(CAT_COLS)

print("\nStatic numeric columns:")
print(STATIC_NUM_COLS)

# =========================
# CHECK TARGET EXISTENCE
# =========================
print("\nTarget columns exist in train:")
print({col: col in train.columns for col in TARGET_COLS})

print("\nTarget columns exist in test:")
print({col: col in test.columns for col in TARGET_COLS})
```

```python
# =========================
# CELL 3 — Static Numeric Missing Value Handling
# Safest version: missing flags + train median imputation
# =========================

# Copy before imputation for rollback
train_before_static_impute = train.copy()
test_before_static_impute = test.copy()

# Static numeric columns that may contain missing values
STATIC_NUM_COLS = [
    "is_home",
    "neutral",
    "population_team",
    "population_opp",
    "gdp_per_capita_team",
    "gdp_per_capita_opp",
    "altitude_venue",
    "distance_travel_team",
    "distance_travel_opp",
    "temperature_venue"
]

STATIC_NUM_COLS = [
    col for col in STATIC_NUM_COLS
    if col in train.columns and col in test.columns
]

# Add missing indicator flags
STATIC_MISSING_FLAG_COLS = []

for col in STATIC_NUM_COLS:
    flag_col = f"{col}_missing"
    train[flag_col] = train[col].isna().astype(int)
    test[flag_col] = test[col].isna().astype(int)
    STATIC_MISSING_FLAG_COLS.append(flag_col)

# Fill missing values using TRAIN median only
STATIC_IMPUTE_VALUES = {}

for col in STATIC_NUM_COLS:
    median_val = train[col].median()
    STATIC_IMPUTE_VALUES[col] = median_val

    train[col] = train[col].fillna(median_val)
    test[col] = test[col].fillna(median_val)

# Add missingness intensity feature
train["static_missing_count"] = train[STATIC_MISSING_FLAG_COLS].sum(axis=1)
test["static_missing_count"] = test[STATIC_MISSING_FLAG_COLS].sum(axis=1)

# =========================
# VALIDATION CHECK
# =========================
print("Static numeric columns:")
print(STATIC_NUM_COLS)

print("\nMissing flag columns:")
print(STATIC_MISSING_FLAG_COLS)

print("\nStatic impute values from train median:")
for k, v in STATIC_IMPUTE_VALUES.items():
    print(f"{k}: {v}")

print("\nRemaining missing values in train static numeric:")
display(train[STATIC_NUM_COLS].isna().sum().to_frame("missing_count"))

print("\nRemaining missing values in test static numeric:")
display(test[STATIC_NUM_COLS].isna().sum().to_frame("missing_count"))

print("\nStatic missing count distribution — train:")
display(train["static_missing_count"].value_counts().sort_index().to_frame("count"))

print("\nStatic missing count distribution — test:")
display(test["static_missing_count"].value_counts().sort_index().to_frame("count"))

# Hard safety check
assert train[STATIC_NUM_COLS].isna().sum().sum() == 0
assert test[STATIC_NUM_COLS].isna().sum().sum() == 0

print("\n✅ Cell 3 done: static numeric missing values handled safely.")
```

```python
# =========================
# CELL 4 — Audit Train-Test Column Mismatch
# =========================

train_cols = set(train.columns)
test_cols = set(test.columns)

cols_only_in_train = sorted(list(train_cols - test_cols))
cols_only_in_test = sorted(list(test_cols - train_cols))
cols_in_both = sorted(list(train_cols & test_cols))

print("Columns only in train:")
print(cols_only_in_train)

print("\nColumns only in test:")
print(cols_only_in_test)

print("\nNumber of shared columns:", len(cols_in_both))

# =========================
# Candidate historical columns from train
# These cannot be used directly unless reconstructed for test
# =========================

HISTORICAL_CANDIDATE_COLS = [
    "elo_team",
    "elo_opponent",
    "elo_diff",

    "rank_team",
    "rank_opponent",
    "rank_diff",

    "team_points_last5",
    "opp_points_last5",
    "points_last5_diff",

    "team_points_last10",
    "opp_points_last10",

    "team_gd_last5",
    "opp_gd_last5",
    "gd_last5_diff",

    "team_avg_goals_last5",
    "opp_avg_goals_last5",

    "team_avg_conceded_last5",
    "opp_avg_conceded_last5",

    "team_win_rate_last10",
    "opp_win_rate_last10",

    "days_since_last_match_team",
    "days_since_last_match_opp",

    "h2h_points_last5",
    "h2h_gd_last5",
]

HISTORICAL_CANDIDATE_COLS = [
    col for col in HISTORICAL_CANDIDATE_COLS
    if col in train.columns
]

print("\nHistorical candidate columns available in train:")
print(HISTORICAL_CANDIDATE_COLS)

print("\nMissing values in train historical candidate columns:")
display(
    train[HISTORICAL_CANDIDATE_COLS]
    .isna()
    .sum()
    .sort_values(ascending=False)
    .to_frame("missing_count")
    .assign(missing_pct=lambda x: x["missing_count"] / len(train) * 100)
)

# =========================
# Shared base features available in train and test
# =========================

BASE_SHARED_FEATURES = CAT_COLS + STATIC_NUM_COLS + STATIC_MISSING_FLAG_COLS + ["static_missing_count"]

BASE_SHARED_FEATURES = [
    col for col in BASE_SHARED_FEATURES
    if col in train.columns and col in test.columns
]

print("\nBase shared features usable right now:")
print(BASE_SHARED_FEATURES)

print("\nNumber of base shared features:", len(BASE_SHARED_FEATURES))
```

```python
# =========================
# CELL 5A — Temporal Validation Split + CatBoost Baseline
# =========================

from catboost import CatBoostRegressor
from sklearn.metrics import mean_absolute_error

# =========================
# TEMPORAL SPLIT
# =========================
VAL_START_YEAR = 2008

train_df = train[train["year"] < VAL_START_YEAR].copy()
val_df = train[train["year"] >= VAL_START_YEAR].copy()

print("Train period:", train_df["date"].min(), "to", train_df["date"].max())
print("Val period  :", val_df["date"].min(), "to", val_df["date"].max())

print("\nTrain rows:", len(train_df))
print("Val rows  :", len(val_df))

# =========================
# FEATURES
# =========================
FEATURES_BASE = BASE_SHARED_FEATURES.copy()

for bad_col in ["id", "match_id", "date", "team_goals", "opp_goals"]:
    if bad_col in FEATURES_BASE:
        FEATURES_BASE.remove(bad_col)

CAT_FEATURES_BASE = [
    col for col in CAT_COLS
    if col in FEATURES_BASE
]

print("\nNumber of base features:", len(FEATURES_BASE))
print("Categorical features:", CAT_FEATURES_BASE)

# =========================
# DATA MATRICES
# =========================
X_train = train_df[FEATURES_BASE].copy()
X_val = val_df[FEATURES_BASE].copy()

y_train_team = train_df["team_goals"]
y_train_opp = train_df["opp_goals"]

y_val_team = val_df["team_goals"]
y_val_opp = val_df["opp_goals"]

# CatBoost wants categorical as string if mixed/object
for col in CAT_FEATURES_BASE:
    X_train[col] = X_train[col].astype(str)
    X_val[col] = X_val[col].astype(str)

# =========================
# MODEL CONFIG
# =========================
cat_params = {
    "loss_function": "MAE",
    "eval_metric": "MAE",
    "iterations": 1200,
    "learning_rate": 0.03,
    "depth": 6,
    "l2_leaf_reg": 8,
    "random_seed": 42,
    "verbose": 100,
    "allow_writing_files": False
}

# =========================
# TRAIN TEAM GOALS MODEL
# =========================
model_team_cat_base = CatBoostRegressor(**cat_params)

model_team_cat_base.fit(
    X_train,
    y_train_team,
    eval_set=(X_val, y_val_team),
    cat_features=CAT_FEATURES_BASE,
    use_best_model=True
)

# =========================
# TRAIN OPP GOALS MODEL
# =========================
model_opp_cat_base = CatBoostRegressor(**cat_params)

model_opp_cat_base.fit(
    X_train,
    y_train_opp,
    eval_set=(X_val, y_val_opp),
    cat_features=CAT_FEATURES_BASE,
    use_best_model=True
)

# =========================
# VALIDATION PREDICTION
# =========================
val_pred_team_cat_base = model_team_cat_base.predict(X_val)
val_pred_opp_cat_base = model_opp_cat_base.predict(X_val)

val_pred_team_cat_base = np.clip(val_pred_team_cat_base, 0, None)
val_pred_opp_cat_base = np.clip(val_pred_opp_cat_base, 0, None)

mae_team_cat_base = mean_absolute_error(y_val_team, val_pred_team_cat_base)
mae_opp_cat_base = mean_absolute_error(y_val_opp, val_pred_opp_cat_base)
mae_avg_cat_base = (mae_team_cat_base + mae_opp_cat_base) / 2

print("\n=== CatBoost Base Validation MAE ===")
print("Team goals MAE:", mae_team_cat_base)
print("Opp goals MAE :", mae_opp_cat_base)
print("Average MAE   :", mae_avg_cat_base)
```
